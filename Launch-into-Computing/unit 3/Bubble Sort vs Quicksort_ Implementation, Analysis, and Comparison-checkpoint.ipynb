```python
# Bubble Sort vs Quicksort: Implementation, Analysis, and Comparison
### MSc Artificial Intelligence — Case study  
 

###This notebook implements Bubble Sort and Quicksort in Python, analyses their time complexity, compares their performance, 
## and visualises execution time across different input sizes.  
```

```python
import matplotlib.pyplot as plt
```

```python
def bubble_sort(arr):
    n = len(arr)                     # Get the length of the list
    for i in range(n):               # Repeat the process n times
        for j in range(0, n - 1):    # Loop through each pair of neighbours
            if arr[j] > arr[j + 1]:  # Compare the two numbers
                arr[j], arr[j + 1] = arr[j + 1], arr[j]  # Swap them
    return arr                       # Return the sorted list

# Example list to sort
example_list = [5, 3, 8, 1, 4]

# Print the sorted list
print("Bubble Sort Output:", bubble_sort(example_list.copy()))
```

```python
def quicksort(arr):
    if len(arr) <= 1:                # Base case
        return arr

    pivot = arr[len(arr)//2]         # Choose pivot

    left = [x for x in arr if x < pivot]     # Smaller than pivot
    middle = [x for x in arr if x == pivot]  # Equal to pivot
    right = [x for x in arr if x > pivot]    # Greater than pivot

    return quicksort(left) + middle + quicksort(right)

# Example list to sort
example_list = [5, 3, 8, 1, 4]

# Print the sorted list
print("Quicksort Output:", quicksort(example_list.copy()))
```

```python
import time
import random
import matplotlib.pyplot as plt

sizes = [100, 500, 1000, 2000, 5000]
bubble_times = []
quick_times = []

for size in sizes:
    print(f"\n--- Sorting list of size {size} ---")

    data = [random.randint(1, 10000) for _ in range(size)]

    # Bubble Sort
    start = time.time()
    bubble_sorted = bubble_sort(data.copy())
    bubble_times.append(time.time() - start)
    print(f"Bubble Sort (first 10 items): {bubble_sorted[:10]}")

    # Quicksort
    start = time.time()
    quick_sorted = quicksort(data.copy())
    quick_times.append(time.time() - start)
    print(f"Quicksort (first 10 items): {quick_sorted[:10]}")


```

```python
## Time Complexity Analysis

### Bubble Sort
# Worst case: **O(n²)**  
# Reason: Nested loops comparing every pair  
# Poor performance on large datasets  

### Quicksort
# Average case: **O(n log n)**  
# Reason: Divide‑and‑conquer partitioning  
#Excellent performance on medium/large datasets  
```

```python
# graph
plt.plot(sizes, bubble_times, label="Bubble Sort")
plt.plot(sizes, quick_times, label="Quicksort")
plt.xlabel("Input Size")
plt.ylabel("Time (seconds)")
plt.title("Bubble Sort vs Quicksort Performance")
plt.legend()
plt.show()
```
